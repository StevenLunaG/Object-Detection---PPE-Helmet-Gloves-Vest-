```python
import cv2
from ultralytics import YOLO

# 1. Cargar el modelo entrenado
model = YOLO('best.pt') 
```

```python
# 2. Iniciar la captura de video
# El '0' suele ser la cámara web integrada de la laptop. 
# Si tienes una cámara USB externa conectada, podría ser '1' o '2'.
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("❌ Error: No se pudo acceder a la cámara web.")
    exit()

print("✅ Cámara activada para el Portal de Seguridad. Presiona 'q' para salir.")
# 3. Bucle infinito para procesar el video en tiempo real
while True:
    # Leer el frame actual de la cámara
    exito, frame = cap.read()
    
    if not exito:
        print("❌ Error al capturar el frame de video.")
        break

    # 4. Hacer la predicción con YOLOv11
    # conf=0.5 significa que solo mostrará detecciones con más de 50% de confianza
    resultados = model.predict(source=frame, conf=0.5, verbose=False)

    # 5. Dibujar las cajas de los cascos, guantes y chalecos sobre el frame
    # resultados[0].plot() toma la imagen original y le "pinta" las detecciones
    frame_anotado = resultados[0].plot()

    # 6. Mostrar el resultado en una ventana emergente
    cv2.imshow('Portal de Seguridad - Verificación EPP en Vivo', frame_anotado)

    # 7. Condición de salida de emergencia
    # El programa se cerrará si presionas la tecla 'q' en tu teclado
    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("🛑 Cerrando el portal de monitoreo...")
        break

# 8. Liberar los recursos de hardware
cap.release()
cv2.destroyAllWindows()
```
